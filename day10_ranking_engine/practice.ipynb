```python
import pandas as pd 

df = pd.DataFrame({
    "region":["East","East","East","West","West"],
    "product":["Laptop","Phone","Tablet","Laptop","Phone"],
    "revenue":[900,700,700,1000,600]
}
)
```

```python
df.head()
```

```python
df["revenue_rank"] = df.groupby("region")["revenue"].rank(method = "dense",ascending  = False)
df.head()
```

```python
df["region_avg_revenue"] = df.groupby("region")["revenue"].transform("mean")
df.head()
```

```python
df["region_total_revenue"] = df.groupby("region")["revenue"].transform("sum")
df.head()
```

```python
df.drop(columns= "revenue_contribution_pct")
```

```python
df["revenue_contribution_pct"] = df["revenue"]/df["region_total_revenue"] * 100
df.head()
```

```python
df.head()
```

```python
top_product = df[df["revenue_rank"] == 1.0]
top_product.head()
```
