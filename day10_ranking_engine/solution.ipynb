```python
import pandas as pd 
import numpy as np

df= pd.read_csv("product_sales.csv")
df.head()
```

```python
df["revenue_rank"] = df.groupby("region")["revenue"].rank(method = "dense",ascending = False)
df.head()
```

```python
df["region_avg_revenue"] = df.groupby("region")["revenue"].transform("mean")
df.head()
```

```python
df
```

```python
df["region_total_revenue"] = df.groupby("region")["revenue"].transform("sum")
df.head()
```

```python
df["revenue_contribution_pct"] = df["revenue"]/ df["region_total_revenue"] * 100
df.head()
```

```python
df["above_region_avg"] = np.where(df["revenue"] > df["region_avg_revenue"],"YES","NO")
df.head()
```

```python
top_2_products = df[df["revenue_rank"] <= 2]
top_2_products
```

```python
top_2_products = top_2_products.sort_values(by= ["region","revenue"],ascending = [True,False])
top_2_products.head()
```
