```python
import pandas as pd 

orders_df = pd.read_csv("orders.csv")
orders_df
```

```python
payments_df = pd.read_csv("payments.csv")
payments_df.head()
```

```python
matching_records = orders_df.merge(payments_df,on = "order_id",how = "inner")
matching_records
```

```python
all_orders_df = orders_df.merge(payments_df,on = "order_id", how = "left")
all_orders_df
```

```python
recon = orders_df.merge(payments_df,on= "order_id",how="outer",indicator = True)
print(recon)
```

```python
recon["record_source"] = recon["_merge"].map({
    "both":"orders & payments",
    "left_only":"orders only",
    "right_only":"payments only"
})
print(recon.head())
```

```python
print(recon[["order_id","record_source"]])
```

```python
exactly_one_record  = orders_df.merge(payments_df, on= "order_id",how= "left",
                                      validate = "one_to_one",suffixes=("_orders","_payments"))
exactly_one_record
```
