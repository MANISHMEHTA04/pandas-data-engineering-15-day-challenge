```python
import pandas as pd 

df = pd.read_csv("regional_sales.csv")
df.head()
```

```python
df.info()
```

```python
pivot_report = df.pivot_table(
    index = "region",
    columns= "product",
    values="revenue",
    aggfunc= "sum"
)
pivot_report.head()
```

```python
month_report = df.pivot_table(
    index = "region",
    columns= "month",
    values = "revenue",
    aggfunc= "sum"
)
month_report
```

```python
df
```

```python

month_report = month_report.reset_index()
long_month_report = month_report.melt(
    id_vars = "region",
    value_vars = ["Jan","Feb"],
    var_name = "month",
    value_name = "total_revenue"
)
long_month_report
```
