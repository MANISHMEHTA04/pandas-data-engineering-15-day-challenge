```python
import pandas as pd

data = {
    "customer_id": ["C101","C101","C101","C102","C102","C102"],
    "month": ["Jan","Feb","Mar","Jan","Feb","Mar"],
    "revenue": [1000,1200,1500,2000,1800,2200]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
df["previous_revenue"] = df.groupby("customer_id")["revenue"].shift(1)
df
```

```python
df["revenue_change"] = df.groupby("customer_id")["revenue"].diff()
df
```

```python
df =df.drop(columns= "difference")
df
```

```python
df["revenue_pct_change"] = df.groupby("customer_id")["revenue"].pct_change() * 100 
df
```

```python
df["cumulative_revenue"] = df.groupby("customer_id")["revenue"].cumsum()
df
```

```python
df["month_sequence"] = df.groupby("customer_id").cumcount() + 1
df
```
