```python
delivered_df = df[df["shipment_status"] == "DELIVERED"]
delivered_df.head()
```

```python
warehouse_kpi = delivered_df.groupby("warehouse").agg(
    shipment_count = ("shipment_id","count"),
    total_revenue = ("revenue","sum"),
    average_revenue = ("revenue","mean"),
    average_delivery_days = ("delivery_days","mean"),
    unique_customers = ("customer_id","nunique")
).reset_index()
```

```python
warehouse_kpi.head()
```

```python
warehouse_kpi.nlargest(1,"average_delivery_days")
```

```python
warehouse_kpi[warehouse_kpi["total_revenue"] > 20000]
```

```python
warehouse_kpi = warehouse_kpi.sort_values("total_revenue", ascending=False)
```
