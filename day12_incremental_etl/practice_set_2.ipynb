```python
import pandas as pd 

target_df = pd.DataFrame(
    {
        "customer_id":["101","102"],
        "city":["Pune","Mumbai"],
        "updated_at" : ["2026-09-20","2026-09-21"]
    }
)

batch_df = pd.DataFrame(
    {
        "customer_id" : ["102","103"],
        "city":["Delhi","Mumbai"],
        "updated_at":["2026-09-29","2026-09-29"]
    }
)

```

```python
target_df.dtypes
```

```python
target_df["updated_at"] = pd.to_datetime(target_df["updated_at"],errors = "coerce")
```

```python
batch_df["updated_at"] = pd.to_datetime(batch_df["updated_at"],errors = "coerce")
```

```python
combined_df = pd.concat([target_df,batch_df],ignore_index = True)
combined_df
```

```python
combined_df = combined_df.sort_values(by=["customer_id","updated_at"],ascending=[True,False])
```

```python
combined_df = combined_df.drop_duplicates(subset = ["customer_id"],keep = "first")
combined_df
```
