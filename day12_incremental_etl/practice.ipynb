```python
import pandas as pd

target_df = pd.DataFrame(
    {
        "customer_id":[101,102,103],
        "name" : ["Amit","Neha","Raj"]
    }
)

batch_df = pd.DataFrame(
    {
        "customer_id":[103,104,105],
        "name":["Raj","Priya","Sam"]
    }
)
 


```

```python
target_df
```

```python
batch_df
```

```python
new_df = batch_df.merge(target_df,on = "customer_id",how = "left",indicator= True)
new_df 
```

```python
new_df = new_df[new_df["_merge"] == "left_only"]
```
