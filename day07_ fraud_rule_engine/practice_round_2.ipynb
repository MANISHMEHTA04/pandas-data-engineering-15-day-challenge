```python
import pandas as pd
import numpy as np 

df = pd.DataFrame({
    "transaction_id": [2001, 2002, 2003, 2004, 2005, 2006],
    "amount": [5000, 65000, 25000, 90000, 15000, 55000],
    "country": ["India", "USA", "UK", "Russia", "India", "Germany"],
    "channel": ["UPI", "CARD", "CARD", "WIRE", "UPI", "WIRE"]
})
```

```python
df.head()
```

```python
country_risk = {
    "India": "LOW",
    "USA": "MEDIUM",
    "UK": "MEDIUM",
    "Russia": "HIGH",
    "Germany": "LOW"
}

df["country_risk"] = df["country"].map(country_risk)
df
```

```python
df["high_value_flag"] = np.where(df["amount"] >= 50000,"YES","NO")
df
```

```python
df = df.assign(
    processing_fee = df["amount"] * 0.02
)
df
```

```python
def risk_rule(row):
    if row["amount"] >= 50000 and row["country_risk"] == "HIGH":
        return "HIGH"
    elif row["amount"] >= 50000 or row["channel"] == "WIRE":
        return "MEDIUM"
    else:
        return "LOW"

df["risk_level"] = df.apply(risk_rule, axis = 1)
df
```
