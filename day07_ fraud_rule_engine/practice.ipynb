```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "transaction_id": [1001, 1002, 1003, 1004, 1005],
    "amount": [2500, 18000, 75000, 1200, 45000],
    "country": ["India", "USA", "India", "UK", "Germany"]
})

df
```

```python
df["transaction_fees"] = df["amount"] * 1.5/100
df
```

```python
df["category"] = np.where(df["amount"] >= 40000,"HIGH","NORMAL")
df
```

```python
df["international_flag"] = np.where(df["country"] != "India","YES","NO")
df
```

```python
df["high_international_flag"] = np.where((df["amount"] >= 40000) & (df["country"] != "India"),"REVIEW","OK")
df
```
