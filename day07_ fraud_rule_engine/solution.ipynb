```python
import pandas as pd
import numpy as np 

country_risk = {
    "India": "LOW",
    "USA": "MEDIUM",
    "UK": "MEDIUM",
    "Russia": "HIGH",
    "Germany": "LOW"
}

```

```python
df = pd.read_csv("fraud_transactions.csv")
df.head()
```

```python
df["country_risk"] = df["country"].map(country_risk)
df
```

```python
df["high_value_flag"] = np.where(df["amount"] >= 50000,"YES","NO")
df
```

```python
df["transaction_fee"] = df["amount"] * 1.5 / 100
df
```

```python
def risk_level(row):
    if row["amount"] >= 75000 and row["country_risk"] == "HIGH" and row["status"] == "SUCCESS":
        return "HIGH"
    elif row["amount"] >= 50000 or row["channel"] == "WIRE":
        return "MEDIUM"
    else:
        return "LOW"

df["risk_level"] = df.apply(risk_level,axis= 1)
df
```

```python
df["manual_review"] = np.where(df["risk_level"] == "HIGH","YES","NO")
df
```

```python
df = df[["transaction_id","customer_id","amount","country","country_risk","risk_level","manual_review"]]
```
