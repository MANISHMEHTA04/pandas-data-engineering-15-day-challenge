```python
import pandas as pd 

df = pd.read_json("nested_orders.json")
df["customer"].head()
```

```python
customer_df = pd.json_normalize(df["customer"])
customer_df.head()
```

```python
combined_df = pd.concat([df,customer_df],axis = 1)
```

```python
combined_df
```

```python
item_df = combined_df.explode("items")
item_df.head()
```

```python
item_df.shape
```

```python
item_df[["order_id","items"]]
```

```python
item_df = item_df.dropna(subset= ["items"])
```

```python
item_df
```

```python
items_df = pd.json_normalize(item_df["items"])
```

```python
items_df
```

```python
item_df.index
```

```python
items_df.index
```

```python
item_df.reset_index(drop= True)
```

```python
items_df.reset_index(drop= True)
```

```python
final_df = pd.concat([item_df,items_df],axis = 1)
final_df
```

```python
final_df = final_df.drop(columns= ["customer","items"])
```

```python
final_df = final_df.assign(
    item_revenue = final_df["quantity"] * final_df["price"]
)
```

```python
final_df
```

```python
item_total_revenue = final_df.groupby("order_id")["item_revenue"].sum()
```

```python
item_total_revenue
```

```python
item_revenue_per_city = final_df.groupby("city")["item_revenue"].sum()
```

```python
item_revenue_per_city
```

```python
product_total_revenue = final_df.groupby("product_name")["item_revenue"].sum()
```

```python
highest_product_revenue = product_total_revenue.nlargest(1)
```

```python
highest_product_revenue
```
